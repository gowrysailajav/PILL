```python
import os
import h5py
import numpy as np
import torch
from torch.utils.data import Dataset


class Landslide4SenseDataset(Dataset):
    """
    Landslide4Sense HDF5 dataset.

    Input:
        14 channels
        B0-B11 : Sentinel-2 bands
        B12    : Slope
        B13    : DEM

    Output:
        image : (14, 128, 128), float32
        mask  : (128, 128), long
    """
    BAND_MEAN = np.array([
        0.3607, 0.3782, 0.3782, 0.3963,
        0.4279, 0.4820, 0.5111, 0.5070,
        0.5348, 0.1868, 0.4058, 0.2813,
        0.0509, 0.3913
    ], dtype=np.float32)

    BAND_STD = np.array([
        0.1967, 0.2019, 0.2019, 0.2078,
        0.2143, 0.2278, 0.2354, 0.2338,
        0.2437, 0.0942, 0.2147, 0.1654,
        0.0334, 0.1944
    ], dtype=np.float32)

    def __init__(
        self,
        data_root: str,
        split: str = "train",
        augment: bool = True
    ):

        assert split in ("train", "val")

        self.split = split
        self.augment = augment and (split == "train")

        # --------------------------------------------------
        # Dataset directories
        # --------------------------------------------------

        folder_map = {
            "train": "TrainData",
            "val": "ValidData"
        }

        folder = os.path.join(
            data_root,
            folder_map[split]
        )

        self.img_dir = os.path.join(folder, "img")
        self.mask_dir = os.path.join(folder, "mask")

        # --------------------------------------------------
        # Check directories
        # --------------------------------------------------

        if not os.path.isdir(self.img_dir):
            raise FileNotFoundError(
                f"Image directory not found:\n{self.img_dir}"
            )

        if not os.path.isdir(self.mask_dir):
            raise FileNotFoundError(
                f"Mask directory not found:\n{self.mask_dir}"
            )

        # --------------------------------------------------
        # Find image files
        # --------------------------------------------------

        self.img_files = sorted(
            [
                f for f in os.listdir(self.img_dir)
                if f.lower().endswith(".h5")
            ]
        )

        if len(self.img_files) == 0:
            raise RuntimeError(
                f"No HDF5 images found in:\n{self.img_dir}"
            )

        # --------------------------------------------------
        # Create image-mask pairs
        # --------------------------------------------------

        self.samples = []

        for img_file in self.img_files:

            # Case 1:
            # image_1.h5 -> mask_1.h5
            if img_file.startswith("image_"):
                mask_file = img_file.replace(
                    "image_", "mask_", 1
                )

            # Case 2:
            # otherwise use same filename
            # e.g. 1.h5 -> 1.h5
            else:
                mask_file = img_file

            mask_path = os.path.join(
                self.mask_dir,
                mask_file
            )

            if os.path.isfile(mask_path):

                self.samples.append(
                    (img_file, mask_file)
                )

        # --------------------------------------------------
        # Verify matching files
        # --------------------------------------------------

        if len(self.samples) == 0:

            raise RuntimeError(
                "\nNo image-mask pairs could be created.\n"
                f"Image directory: {self.img_dir}\n"
                f"Mask directory : {self.mask_dir}\n"
                f"Number of images: {len(self.img_files)}\n"
                "\nCheck the filenames in the image and mask folders."
            )

        if len(self.samples) != len(self.img_files):

            print(
                f"WARNING: {len(self.img_files)} images found, "
                f"but only {len(self.samples)} have matching masks."
            )

        print(
            f"  [{split}] Images : {len(self.img_files)}"
        )

        print(
            f"  [{split}] Pairs  : {len(self.samples)}"
        )

    # ------------------------------------------------------
    # Dataset length
    # ------------------------------------------------------

    def __len__(self):
        return len(self.samples)

    # ------------------------------------------------------
    # HDF5 loader
    # ------------------------------------------------------

    def _load_h5(self, path):

        with h5py.File(path, "r") as f:

            key = list(f.keys())[0]

            data = f[key][:]

        return data.astype(np.float32)

    # ------------------------------------------------------
    # Get sample
    # ------------------------------------------------------

    def __getitem__(self, idx):

        img_file, mask_file = self.samples[idx]

        # ==================================================
        # Load image
        # ==================================================

        img_path = os.path.join(
            self.img_dir,
            img_file
        )

        img = self._load_h5(img_path)

        # --------------------------------------------------
        # Ensure (C,H,W)
        # --------------------------------------------------

        if img.ndim == 3:

            # H,W,C
            if img.shape[-1] == 14:

                img = img.transpose(2, 0, 1)

            # C,H,W already
            elif img.shape[0] == 14:

                pass

            else:

                raise ValueError(
                    f"Unexpected image shape "
                    f"{img.shape} in {img_file}"
                )

        else:

            raise ValueError(
                f"Unexpected image dimensions "
                f"{img.shape} in {img_file}"
            )

        # ==================================================
        # Normalize bands
        # ==================================================

        mean = self.BAND_MEAN[:, None, None]
        std = self.BAND_STD[:, None, None]

        img = (
            img - mean
        ) / (
            std + 1e-8
        )

        img_tensor = torch.from_numpy(
            img
        ).float()

        # ==================================================
        # Load mask
        # ==================================================

        mask_path = os.path.join(
            self.mask_dir,
            mask_file
        )

        mask = self._load_h5(mask_path)

        # --------------------------------------------------
        # Remove singleton dimensions
        # --------------------------------------------------

        mask = np.squeeze(mask)

        if mask.shape != (128, 128):

            raise ValueError(
                f"Unexpected mask shape "
                f"{mask.shape} in {mask_file}"
            )

        # --------------------------------------------------
        # Ensure binary mask
        # --------------------------------------------------

        mask = (mask > 0).astype(np.int64)

        mask_tensor = torch.from_numpy(
            mask
        ).long()

        # ==================================================
        # Data augmentation
        # ==================================================

        if self.augment:

            # Horizontal flip
            if torch.rand(1).item() > 0.5:

                img_tensor = torch.flip(
                    img_tensor,
                    dims=[2]
                )

                mask_tensor = torch.flip(
                    mask_tensor,
                    dims=[1]
                )

            # Vertical flip
            if torch.rand(1).item() > 0.5:

                img_tensor = torch.flip(
                    img_tensor,
                    dims=[1]
                )

                mask_tensor = torch.flip(
                    mask_tensor,
                    dims=[0]
                )

        return img_tensor, mask_tensor
```

```python
# Shared loss and metrics function definition


class DiceBCELoss(nn.Module):
    """
    Combined Dice + Binary Cross-Entropy Loss
    """

    def __init__(self, pos_weight=5.0, dice_w=0.5):
        super().__init__()
        self.dice_w = dice_w
        self.bce_w  = 1.0 - dice_w
        self.pw = pos_weight

    def _dice(self, logits, targets, smooth=1.0):
        p = torch.sigmoid(logits)
        if p.dim() == 4:
            p = p.squeeze(1)  # (B,1,H,W) → (B,H,W)
        if targets.dim() == 4:
            targets = targets.squeeze(1)
        
        num = 2.0*(p*targets).sum(dim=(1,2)) + smooth
        denom = p.sum(dim=(1,2)) + targets.sum(dim=(1,2)) + smooth
        return (1.0 - num / denom).mean()  # average over batch

    def forward(self, logits, targets):
        if logits.dim() == 4:
            logits = logits.squeeze(1)  # (B,1,H,W) → (B,H,W)
        if targets.dim() == 4:
            targets = targets.squeeze(1)

        pw = torch.tensor([self.pw], device=logits.device)
        bce = F.binary_cross_entropy_with_logits(logits, targets.float(), pos_weight=pw)

        return (self.bce_w * bce + self.dice_w * self._dice(logits, targets))

    @torch.no_grad()
    def evaluate(model, loader, device, name='Model'):
        model.eval()
        all_probs = []
        all_preds = []
        all_labels = []

        for imgs, masks in loader:
            imgs = imgs.to(device)
            out = model(imgs)

            if isinstance(out, (list, tuple)):
                out = out[0]  # handle models that return (logits, aux)

            prob = (torch.sigmoid(out.squeeze(1)).cpu().numpy().reshape(-1))
            pred = (prob > 0.5).astype(np.int32)
            lbl = masks.numpy().astype(np.int32).reshape(-1)

            all_probs.append(prob)
            all_preds.append(pred)
            all_labels.append(lbl)

        probs = np.concatenate(all_probs)
        preds = np.concatenate(all_preds)
        labels = np.concatenate(all_labels)

        TP = int(((preds == 1) & (labels == 1)).sum())
        FP = int(((preds == 1) & (labels == 0)).sum())
        TN = int(((preds == 0) & (labels == 0)).sum())
        FN = int(((preds == 0) & (labels == 1)).sum())

        e = 1e-8

        acc = (TP + TN) / (TP + TN + FP + FN + e)
        prec = TP / (TP + FP + e)
        rec = TP / (TP + FN + e)
        f1 = 2 * (prec * rec) / (prec + rec + e)
        spec = TN / (TN + FP + e)
        iou = TP / (TP + FP + FN + e)
        dice = 2 * TP / (2 * TP + FP + FN + e)
        bacc = (rec + spec) / 2.0

        try:
            auc_roc = roc_auc_score(labels, probs)
            auc_pr  = average_precision_score(labels, probs)
        except Exception:
            auc_roc = auc_pr  = 0.0

        n_p = sum(p.numel() for p in model.parameters() if p.requires_grad)

        res = {
            'Model'      : name,
            'Accuracy'    : round(float(acc), 4),
            'Precision'   : round(float(prec), 4),
            'Recall'      : round(float(rec), 4),
            'F1'          : round(float(f1), 4),
            'Specificity' : round(float(spec), 4),
            'IoU'         : round(float(iou), 4),
            'Dice'        : round(float(dice), 4),
            'Balanced Acc' : round(float(bacc), 4),
            'AUC-ROC'     : round(float(auc_roc), 4),
            'AUC-PR'      : round(float(auc_pr), 4),
            'Params_M'      : round(n_p / 1e6, 3),
            'TP'         : TP,
            'FP'         : FP,
            'TN'         : TN,
            'FN'         : FN,
        }


        print(f"\n{ '='*50 }")
        print(f"\n{name} Evaluation Results for {name} on {len(loader.dataset)} samples ")
        print(f"\n{ '='*50 }")

        for k in ['Accuracy', 'Precision', 'Recall', 'F1', 'Specificity', 'IoU', 'Dice', 'Balanced Acc', 'AUC-ROC', 'AUC-PR']:
            print(f"{k:<12} : {res[k]:.4f}")

        print(f"\n{'TP'} : {res['TP']}")
        print(f"{'FP'} : {res['FP']}")
        print(f"{'TN'} : {res['TN']}")
        print(f"{'FN'} : {res['FN']}")  

        print(f"\n{ '*'*50 }")

        return res
```

```python
import time
import torch


def measure_inference_time(model, sample_input, warmup=10, iterations=100):
    """Measure average forward-pass latency in milliseconds per image."""
    model.eval()
    sample_input = sample_input.to(device)

    with torch.inference_mode():
        # Warm-up
        for _ in range(warmup):
            _ = model(sample_input)

        if sample_input.is_cuda:
            torch.cuda.synchronize()

        start = time.perf_counter()

        for _ in range(iterations):
            _ = model(sample_input)

        if sample_input.is_cuda:
            torch.cuda.synchronize()

        elapsed = time.perf_counter() - start

    # Average latency per image, not per batch
    return (elapsed / iterations / sample_input.shape[0]) * 1000


def measure_flops(model, sample_input):
    """Estimate GFLOPs for one forward pass on the sample input."""
    from torch.profiler import profile, ProfilerActivity

    model.eval()
    sample_input = sample_input.to(device)

    activities = [ProfilerActivity.CPU]
    if sample_input.is_cuda:
        activities.append(ProfilerActivity.CUDA)

    with torch.inference_mode():
        with profile(
            activities=activities,
            with_flops=True,
            record_shapes=True
        ) as prof:
            _ = model(sample_input)

    total_flops = sum(
        event.flops or 0
        for event in prof.key_averages()
    )

    # GFLOPs per input batch; use batch size 1 for per-image GFLOPs
    return total_flops / 1e9



def measure_fps(model, sample_input, warmup=10, iterations=100):
    """
    Measure inference throughput in images per second.
    Uses synchronized timing on CUDA.
    """
    model.eval()
    sample_input = sample_input.to(device)
    batch_size = sample_input.shape[0]

    with torch.inference_mode():
        # Warm-up
        for _ in range(warmup):
            _ = model(sample_input)

        if sample_input.is_cuda:
            torch.cuda.synchronize()

        start = time.perf_counter()

        for _ in range(iterations):
            _ = model(sample_input)

        if sample_input.is_cuda:
            torch.cuda.synchronize()

        elapsed = time.perf_counter() - start

    fps = (iterations * batch_size) / elapsed
    return fps


def measure_peak_gpu_memory(model, sample_input, warmup=5):
    """
    Measure peak GPU memory allocated during inference, in MiB.
    Returns None if CUDA is unavailable.
    """
    if not torch.cuda.is_available() or not sample_input.is_cuda:
        return None

    model.eval()
    sample_input = sample_input.to(device)

    with torch.inference_mode():
        # Warm-up before measuring
        for _ in range(warmup):
            _ = model(sample_input)

        torch.cuda.synchronize()

        # Reset peak tracking after warm-up
        torch.cuda.reset_peak_memory_stats(sample_input.device)

        _ = model(sample_input)
        torch.cuda.synchronize()

        peak_memory_mb = (
            torch.cuda.max_memory_allocated(sample_input.device)
            / (1024 ** 2)
        )

    return peak_memory_mb


def measure_model_size(checkpoint_path):
    """
    Measure saved checkpoint size in MB (decimal megabytes).
    """
    return os.path.getsize(checkpoint_path) / (1024 ** 2)
```

```python

# ============================================================
# METHOD 12: TILL 
# ============================================================
import tensorflow as tf
import numpy as np

from tensorflow.keras import layers, Model

# ── Band-Group Attention ──────────────────────────────────────────────────────
class BandGroupAttention(layers.Layer):
    def __init__(self, out_dim, **kwargs):
        super().__init__(**kwargs)
        self.proj_vis     = layers.Conv2D(out_dim//4, 1, padding='same')
        self.proj_swir    = layers.Conv2D(out_dim//2, 1, padding='same')
        self.proj_terrain = layers.Conv2D(out_dim//4, 1, padding='same')
        self.gap          = layers.GlobalAveragePooling2D()
        self.ca_vis       = layers.Dense(out_dim//4,  activation='sigmoid')
        self.ca_swir      = layers.Dense(out_dim//2,  activation='sigmoid')
        self.ca_terr      = layers.Dense(out_dim//4,  activation='sigmoid')
        self.fuse         = layers.Conv2D(out_dim, 1, padding='same')
        self.norm         = layers.LayerNormalization()
        self.act          = layers.Activation('gelu')

    def call(self, x, training=False):
        vis     = x[:,:,:, 0:4]
        swir    = x[:,:,:, 4:12]
        terrain = x[:,:,:,12:14]

        pv = self.proj_vis(vis)
        ps = self.proj_swir(swir)
        pt = self.proj_terrain(terrain)

        B = tf.shape(x)[0]
        av = tf.reshape(self.ca_vis( self.gap(pv)), (B,1,1,-1))
        as_= tf.reshape(self.ca_swir(self.gap(ps)), (B,1,1,-1))
        at = tf.reshape(self.ca_terr(self.gap(pt)), (B,1,1,-1))

        pv = pv * av
        ps = ps * as_
        pt = pt * at

        out = tf.concat([pv, ps, pt], axis=-1)
        return self.act(self.norm(self.fuse(out)))


# ── Terrain Gate ──────────────────────────────────────────────────────────────
class TerrainGate(layers.Layer):
    def __init__(self, dim, **kwargs):
        super().__init__(**kwargs)
        self.slope_enc = layers.Conv2D(dim//2, 3, padding='same', activation='relu')
        self.dem_enc   = layers.Conv2D(dim//2, 3, padding='same', activation='relu')
        self.fuse      = layers.Conv2D(dim,    3, padding='same', activation='relu')
        self.gate_conv = layers.Conv2D(1,      1, padding='same', activation='sigmoid')
        self.bn        = layers.BatchNormalization()
        self.residual  = layers.Conv2D(dim,    1, padding='same')
        # Resize layers — created in __init__, used in call
        self.resize_slope = layers.Resizing(0, 0)   # placeholder
        self.resize_dem   = layers.Resizing(0, 0)

    def call(self, x, slope, dem, training=False):
        H = x.shape[1]
        W = x.shape[2]
        # Use bilinear resize via UpSampling or avg pool approach
        # Resize terrain maps to match x using Lambda-safe ops
        slope_r = tf.image.resize(slope, (H, W))
        dem_r   = tf.image.resize(dem,   (H, W))

        s = self.slope_enc(slope_r)
        d = self.dem_enc(dem_r)
        t = self.fuse(tf.concat([s, d], axis=-1))
        g = self.gate_conv(t)

        gated = x * g
        res   = self.residual(x)
        out   = self.bn(gated + res, training=training)
        return out, g


# ── Depthwise Separable Conv Block ────────────────────────────────────────────
class DSConvBlock(layers.Layer):
    def __init__(self, dim, stride=1, **kwargs):
        super().__init__(**kwargs)
        self.dw     = layers.DepthwiseConv2D(3, strides=stride, padding='same')
        self.pw     = layers.Conv2D(dim, 1, padding='same')
        self.bn1    = layers.BatchNormalization()
        self.bn2    = layers.BatchNormalization()
        self.drop   = layers.Dropout(0.05)
        self.res    = layers.Conv2D(dim, 1, strides=stride, padding='same')
        self.res_bn = layers.BatchNormalization()

    def call(self, x, training=False):
        r = self.res_bn(self.res(x), training=training)
        x = tf.nn.gelu(self.bn1(self.dw(x), training=training))
        x = tf.nn.gelu(self.bn2(self.pw(x), training=training))
        x = self.drop(x, training=training)
        return x + r


# ── Multi-Scale Feature Aggregation ──────────────────────────────────────────
class MSFABlock(layers.Layer):
    def __init__(self, dim, **kwargs):
        super().__init__(**kwargs)
        d = max(dim // 4, 8)
        self.c1  = layers.Conv2D(d, 1, padding='same')
        self.c3  = layers.Conv2D(d, 3, padding='same', dilation_rate=1)
        self.c5  = layers.Conv2D(d, 3, padding='same', dilation_rate=3)
        self.c7  = layers.Conv2D(d, 3, padding='same', dilation_rate=5)
        self.bn1 = layers.BatchNormalization()
        self.bn3 = layers.BatchNormalization()
        self.bn5 = layers.BatchNormalization()
        self.bn7 = layers.BatchNormalization()
        self.fuse= layers.Conv2D(dim, 1, padding='same')
        self.bn  = layers.BatchNormalization()

    def call(self, x, training=False):
        r1 = tf.nn.gelu(self.bn1(self.c1(x), training=training))
        r3 = tf.nn.gelu(self.bn3(self.c3(x), training=training))
        r5 = tf.nn.gelu(self.bn5(self.c5(x), training=training))
        r7 = tf.nn.gelu(self.bn7(self.c7(x), training=training))
        out= tf.concat([r1, r3, r5, r7], axis=-1)
        return tf.nn.gelu(self.bn(self.fuse(out), training=training))


# ── Recurrent Refinement Unit ─────────────────────────────────────────────────
class RecurrentRefinement(layers.Layer):
    def __init__(self, dim, T=3, **kwargs):
        super().__init__(**kwargs)
        self.T = T
        self.input_proj   = layers.Conv2D(dim, 1, padding='same')
        self.input_bn     = layers.BatchNormalization()
        self.refine_conv1 = layers.Conv2D(dim,    3, padding='same')
        self.refine_bn1   = layers.BatchNormalization()
        self.refine_conv2 = layers.Conv2D(dim,    3, padding='same')
        self.refine_bn2   = layers.BatchNormalization()
        self.refine_conv3 = layers.Conv2D(dim//2, 3, padding='same')
        self.refine_bn3   = layers.BatchNormalization()
        self.out_conv     = layers.Conv2D(1, 1, activation='sigmoid')
        self.drop         = layers.Dropout(0.1)

    def call(self, features, init_pred, training=False):
        pred = init_pred
        for _ in range(self.T):
            x = tf.concat([features, pred], axis=-1)
            x = tf.nn.gelu(
                self.input_bn(self.input_proj(x), training=training))
            x = tf.nn.gelu(
                self.refine_bn1(self.refine_conv1(x), training=training))
            x = self.drop(x, training=training)
            x = tf.nn.gelu(
                self.refine_bn2(self.refine_conv2(x), training=training))
            x = tf.nn.gelu(
                self.refine_bn3(self.refine_conv3(x), training=training))
            pred = self.out_conv(x)
        return pred


# ── Decoder Block ─────────────────────────────────────────────────────────────
class DecBlock(layers.Layer):
    def __init__(self, filters, **kwargs):
        super().__init__(**kwargs)
        self.up       = layers.UpSampling2D(2, interpolation='bilinear')
        self.skip_proj= layers.Conv2D(filters, 1, padding='same')
        self.skip_bn  = layers.BatchNormalization()
        self.cat_proj = layers.Conv2D(filters, 1, padding='same')
        self.cat_bn   = layers.BatchNormalization()
        self.conv1    = layers.Conv2D(filters, 3, padding='same')
        self.bn1      = layers.BatchNormalization()
        self.conv2    = layers.Conv2D(filters, 3, padding='same')
        self.bn2      = layers.BatchNormalization()
        self.drop     = layers.Dropout(0.1)

    def call(self, x, skip=None, training=False):
        x = self.up(x)
        if skip is not None:
            skip = tf.nn.gelu(
                self.skip_bn(self.skip_proj(skip), training=training))
            x = tf.concat([x, skip], axis=-1)
            x = tf.nn.gelu(
                self.cat_bn(self.cat_proj(x), training=training))
        x = tf.nn.gelu(self.bn1(self.conv1(x), training=training))
        x = self.drop(x, training=training)
        x = tf.nn.gelu(self.bn2(self.conv2(x), training=training))
        return x


# ── Aux Head ──────────────────────────────────────────────────────────────────
class AuxHead(layers.Layer):
    """Auxiliary output head — all ops in call()."""
    def __init__(self, scale, name, **kwargs):
        super().__init__(name=name, **kwargs)
        self.scale   = scale
        self.up      = layers.UpSampling2D(
                            scale, interpolation='bilinear')
        self.conv    = layers.Conv2D(32, 3, padding='same')
        self.bn      = layers.BatchNormalization()
        self.out     = layers.Conv2D(1,  1, activation='sigmoid')

    def call(self, x, training=False):
        x = self.up(x)
        x = tf.nn.gelu(self.bn(self.conv(x), training=training))
        return self.out(x)


# ── Initial Pred Head ─────────────────────────────────────────────────────────
class InitHead(layers.Layer):
    def __init__(self, dim, **kwargs):
        super().__init__(**kwargs)
        self.conv = layers.Conv2D(dim, 3, padding='same')
        self.bn   = layers.BatchNormalization()
        self.out  = layers.Conv2D(1,   1, activation='sigmoid')

    def call(self, x, training=False):
        x = tf.nn.gelu(self.bn(self.conv(x), training=training))
        return x, self.out(x)


# ── Full TerrainFormer ────────────────────────────────────────────────────────
def build_terrainformer(input_shape=(128, 128, 14)):
    inputs = layers.Input(shape=input_shape, name='input')

    # Band-group attention stem
    x     = BandGroupAttention(64, name='bga')(inputs)  # 128x128x64
    skip0 = x

    # Terrain bands extracted inside TerrainGate.call()
    # Pass raw inputs so TerrainGate can slice internally
    slope_inp = inputs[:,:,:,12:13]
    dem_inp   = inputs[:,:,:,13:14]

    # Stage 1: 128→64, C=96
    x           = DSConvBlock(96,  stride=2, name='ds1')(x)
    x           = MSFABlock(96,  name='msfa1')(x)
    x, _        = TerrainGate(96,  name='tg1')(x, slope_inp, dem_inp)
    skip1       = x                                      # 64x64x96

    # Stage 2: 64→32, C=192
    x           = DSConvBlock(192, stride=2, name='ds2')(x)
    x           = MSFABlock(192, name='msfa2')(x)
    x, _        = TerrainGate(192, name='tg2')(x, slope_inp, dem_inp)
    skip2       = x                                      # 32x32x192

    # Stage 3: 32→16, C=320
    x           = DSConvBlock(320, stride=2, name='ds3')(x)
    x           = MSFABlock(320, name='msfa3')(x)
    x, _        = TerrainGate(320, name='tg3')(x, slope_inp, dem_inp)
    skip3       = x                                      # 16x16x320

    # Stage 4: 16→8, C=512 Bottleneck
    x           = DSConvBlock(512, stride=2, name='ds4')(x)
    x           = MSFABlock(512, name='msfa4')(x)
    bottle, _   = TerrainGate(512, name='tg4')(x, slope_inp, dem_inp)
                                                         # 8x8x512

    # Auxiliary output 2 from bottleneck (8→32)
    aux2 = AuxHead(scale=4, name='aux2')(bottle)         # 32x32x1

    # Decoder
    d1   = DecBlock(192, name='dec1')(bottle, skip=skip3) # 16x16
    aux1 = AuxHead(scale=4, name='aux1')(d1)              # 64x64x1

    d2   = DecBlock(96,  name='dec2')(d1, skip=skip2)     # 32x32
    d3   = DecBlock(48,  name='dec3')(d2, skip=skip1)     # 64x64
    d4   = DecBlock(32,  name='dec4')(d3, skip=skip0)     # 128x128

    # Initial prediction + recurrent refinement
    init_feat, init_pred = InitHead(32, name='init_head')(d4)
    rru  = RecurrentRefinement(32, T=3, name='rru')
    main = rru(init_feat, init_pred)                      # 128x128x1

    # Name final output
    main = layers.Conv2D(
        1, 1, activation='sigmoid', name='main')(main)

    return Model(inputs, [main, aux1, aux2],
                 name='TerrainFormer')


model = build_terrainformer()
model.summary()
print(f"\n  Total Parameters : {model.count_params():,}")
print(f"  main : {model.outputs[0].shape}")
print(f"  aux1 : {model.outputs[1].shape}")
print(f"  aux2 : {model.outputs[2].shape}")
print("\n  CELL 3 COMPLETE!")
print("=" * 65)


# ── Focal Loss ────────────────────────────────────────────────────────────────
class FocalLoss(keras.losses.Loss):
    def __init__(self, gamma=2.0, alpha=0.85, **kwargs):
        super().__init__(**kwargs)
        self.gamma = gamma
        self.alpha = alpha

    def call(self, y_true, y_pred):
        yt = tf.cast(y_true, 'float32')
        yp = tf.clip_by_value(
            y_pred,
            clip_value_min=1e-7,
            clip_value_max=1.0-1e-7
        )
        bce = -(
            yt * tf.math.log(yp)
            + (1.0 - yt) * tf.math.log(1.0 - yp)
        )
        pt = yt*yp + (1-yt)*(1-yp)
        at = yt*self.alpha + (1-yt)*(1-self.alpha)
        return tf.reduce_mean(at * tf.math.pow(1-pt, self.gamma) * bce)


# ── Dice Loss ─────────────────────────────────────────────────────────────────
class DiceLoss(keras.losses.Loss):
    def __init__(self, smooth=1.0, **kwargs):
        super().__init__(**kwargs)
        self.smooth = smooth

    def call(self, y_true, y_pred):
        yt    = tf.cast(tf.reshape(y_true, (-1,)), 'float32')
        yp    = tf.reshape(y_pred, (-1,))
        inter = tf.reduce_sum(yt * yp)
        return 1-(2*inter+self.smooth)/(
            tf.reduce_sum(yt)+tf.reduce_sum(yp)+self.smooth)


# ── Tversky Loss ──────────────────────────────────────────────────────────────
class TverskyLoss(keras.losses.Loss):
    """alpha=0.7 penalises FN more — critical for landslide safety."""
    def __init__(self, alpha=0.7, beta=0.3, smooth=1.0, **kwargs):
        super().__init__(**kwargs)
        self.alpha=alpha; self.beta=beta; self.smooth=smooth

    def call(self, y_true, y_pred):
        yt = tf.cast(tf.reshape(y_true, (-1,)), 'float32')
        yp = tf.reshape(y_pred, (-1,))
        TP = tf.reduce_sum(yt*yp)
        FP = tf.reduce_sum((1-yt)*yp)
        FN = tf.reduce_sum(yt*(1-yp))
        return 1-(TP+self.smooth)/(
            TP+self.alpha*FN+self.beta*FP+self.smooth)


# ── TerrainFormer Combined Loss ───────────────────────────────────────────────
class TerrainFormerLoss(keras.losses.Loss):
    """
    0.40 * Focal   (handles extreme class imbalance 2% landslide)
    0.35 * Tversky (penalises missed landslides heavily)
    0.25 * Dice    (optimises overlap quality)
    """
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.focal   = FocalLoss(gamma=2.0, alpha=0.75)
        # self.focal   = FocalLoss(gamma=2.0, alpha=0.85)
        self.tversky = TverskyLoss(alpha=0.7, beta=0.3)
        self.dice    = DiceLoss(smooth=1.0)

    def call(self, y_true, y_pred):
        return (0.35 * self.focal(y_true, y_pred)   +
                0.35 * self.tversky(y_true, y_pred) +
                0.30 * self.dice(y_true, y_pred))


# ── Pixel F1 Metric ───────────────────────────────────────────────────────────
class PixelF1(keras.metrics.Metric):
    def __init__(self, threshold=0.5, **kwargs):
        super().__init__(name='f1', **kwargs)
        self.thr = threshold
        self.tp  = self.add_weight(name='tp', shape=(), initializer='zeros')
        self.fp  = self.add_weight(name='fp', shape=(), initializer='zeros')
        self.fn  = self.add_weight(name='fn', shape=(), initializer='zeros')

    def update_state(self, y_true, y_pred, sample_weight=None):
        yt = tf.cast(tf.reshape(y_true,  (-1,)), 'float32')
        yp = tf.cast(tf.reshape(y_pred,  (-1,)) >= self.thr, 'float32')
        self.tp.assign_add(tf.reduce_sum(yt * yp))
        self.fp.assign_add(tf.reduce_sum((1-yt) * yp))
        self.fn.assign_add(tf.reduce_sum(yt * (1-yp)))

    def result(self):
        p = self.tp/(self.tp+self.fp+1e-7)
        r = self.tp/(self.tp+self.fn+1e-7)
        return 2*p*r/(p+r+1e-7)

    def reset_state(self):
        self.tp.assign(0.); self.fp.assign(0.); self.fn.assign(0.)


# ── IoU Metric ────────────────────────────────────────────────────────────────
class IoUMetric(keras.metrics.Metric):
    def __init__(self, threshold=0.5, **kwargs):
        super().__init__(name='iou', **kwargs)
        self.thr   = threshold
        self.inter = self.add_weight(name='inter', shape=(), initializer='zeros')
        self.union = self.add_weight(name='union', shape=(), initializer='zeros')

    def update_state(self, y_true, y_pred, sample_weight=None):
        yt = tf.cast(tf.reshape(y_true, (-1,)), 'float32')
        yp = tf.cast(tf.reshape(y_pred, (-1,)) >= self.thr, 'float32')
        self.inter.assign_add(tf.reduce_sum(yt*yp))
        self.union.assign_add(tf.reduce_sum(yt)+tf.reduce_sum(yp)-tf.reduce_sum(yt*yp))

    def result(self):
        return (self.inter+1e-7)/(self.union+1e-7)

    def reset_state(self):
        self.inter.assign(0.); self.union.assign(0.)


# ── Dice Metric ───────────────────────────────────────────────────────────────
class DiceMetric(keras.metrics.Metric):
    def __init__(self, threshold=0.5, **kwargs):
        super().__init__(name='dice', **kwargs)
        self.thr   = threshold
        self.inter = self.add_weight(name='inter', shape=(), initializer='zeros')
        self.denom = self.add_weight(name='denom', shape=(), initializer='zeros')

    def update_state(self, y_true, y_pred, sample_weight=None):
        yt = tf.cast(tf.reshape(y_true, (-1,)), 'float32')
        yp = tf.cast(tf.reshape(y_pred, (-1,)) >= self.thr, 'float32')
        self.inter.assign_add(tf.reduce_sum(yt*yp))
        self.denom.assign_add(tf.reduce_sum(yt)+tf.reduce_sum(yp))

    def result(self):
        return (2*self.inter+1)/(self.denom+1)

    def reset_state(self):
        self.inter.assign(0.); self.denom.assign(0.)


# ── LR Schedule ───────────────────────────────────────────────────────────────

class WarmupCosineDecay(
    tf.keras.optimizers.schedules.LearningRateSchedule
):

    def __init__(self, base_lr, total_steps, warmup_steps, min_lr=1e-6):

        super().__init__()

        self.base_lr = float(base_lr)
        self.total_steps = float(total_steps)
        self.warmup_steps = float(warmup_steps)
        self.min_lr = float(min_lr)

    def __call__(self, step):

        step = tf.cast(step, tf.float32)

        warmup = tf.constant(self.warmup_steps, tf.float32)
        total = tf.constant(self.total_steps, tf.float32)

        # Linear warmup
        wlr = self.base_lr * step / tf.maximum(warmup, 1.0)

        # Cosine decay
        progress = (step - warmup) / tf.maximum(total - warmup, 1.0)
        progress = tf.clip_by_value(progress, 0.0, 1.0)

        clr = self.min_lr + 0.5 * (self.base_lr - self.min_lr) * (
            1.0 + tf.cos(np.pi * progress)
        )

        return tf.where(step < warmup, wlr, clr)

    def get_config(self):

        return {
            "base_lr": self.base_lr,
            "total_steps": self.total_steps,
            "warmup_steps": self.warmup_steps,
            "min_lr": self.min_lr
        }





STEPS_PER_EPOCH = len(train_loader)
VAL_STEPS = len(val_loader)
TOTAL_STEPS = EP * STEPS_PER_EPOCH
STEPS = STEPS_PER_EPOCH
WARMUP      = 5 * STEPS


lr_sched  = WarmupCosineDecay(LR, TOTAL_STEPS, WARMUP, min_lr=1e-6)
optimizer = keras.optimizers.AdamW(
    learning_rate=lr_sched,
    weight_decay=1e-4,
    clipnorm=1.0
)

model.compile(
    optimizer    = optimizer,
    loss         = {
        'main': TerrainFormerLoss(),
        'aux1': DiceLoss(),
        'aux2': DiceLoss(),
    },
    loss_weights = {
        'main': 1.0,
        'aux1': 0.4,
        'aux2': 0.2,
    },
    metrics      = {
        'main': [
            keras.metrics.BinaryAccuracy(name='acc', threshold=0.5),
            keras.metrics.AUC(name='auc'),
            PixelF1(threshold=0.3),
            IoUMetric(threshold=0.3),
            DiceMetric(threshold=0.3),
        ]
    }
)

print("  Model compiled!")
print(f"  Epochs      : {EP}")
print(f"  Steps/epoch : {STEPS}")
print(f"  Total steps : {TOTAL_STEPS}")
print(f"  Warmup      : {WARMUP} steps (5 epochs)")
print(f"  Base LR     : {LR}")
print(f"  Min LR      : 1e-6")
print(f"  Optimizer   : AdamW (wd=1e-4, clipnorm=1.0)")
print(f"  Loss main   : 0.40*Focal + 0.35*Tversky + 0.25*Dice")
print(f"  Loss aux1   : Dice (weight=0.4)")
print(f"  Loss aux2   : Dice (weight=0.2)")
print("\n  CELL 4 COMPLETE!")
print("=" * 65)



# ============================================================
# METHOD 12: TILL TRAINING USING SHARED run_training()
# ============================================================

# Use the already-built and compiled TILL model
m12 = model

# Train using the shared run_training() function
m12, r12, h12 = run_training(
    m12,
    'TILL',
    epochs=EP,
    lr=LR
)

# Store results in the common result containers
ALL_RESULTS.append(r12)
ALL_HISTORIES['TILL'] = h12

# Save final best-performing model weights
m12.save_weights(
    CHECKPOINTS_DIR / "Method12_TILL.weights.h5"
)

print(
    f"\n{'='*60}\n"
    f"Method 12 (TILL) training complete.\n"
    f"Model saved successfully.\n"
    f"F1: {r12['F1']} | "
    f"IoU: {r12['IoU']} | "
    f"AUC: {r12['AUC-ROC']}\n"
    f"{'='*60}"
)
```

```python

# ============================================================
# METHOD 11: PILL 
# ============================================================



class DoubleConv(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(in_c, out_c, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_c),
            nn.GELU(),
            nn.Conv2d(out_c, out_c, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_c),
            nn.GELU(),
        )
    def forward(self, x):
        return self.net(x)

class Down(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.pool = nn.MaxPool2d(2)
        self.conv = DoubleConv(in_c, out_c)
    def forward(self, x):
        return self.conv(self.pool(x))

class Up(nn.Module):
    def __init__(self, in_c, out_c):
        super().__init__()
        self.up   = nn.ConvTranspose2d(in_c, in_c // 2, kernel_size=2, stride=2)
        self.conv = DoubleConv(in_c, out_c)
    def forward(self, x, skip):
        x = self.up(x)
        x = torch.cat([skip, x], dim=1)
        return self.conv(x)

class ParameterFieldGenerator(nn.Module):
    def __init__(self, in_channels=14, base_ch=64):
        super().__init__()

        # Encoder
        self.inc = DoubleConv(in_channels, base_ch)        # 14 → 64
        self.d1  = Down(base_ch,     base_ch * 2)          # 64 → 128
        self.d2  = Down(base_ch * 2, base_ch * 4)          # 128 → 256
        self.d3  = Down(base_ch * 4, base_ch * 8)          # 256 → 512
        self.bot = Down(base_ch * 8, base_ch * 16)         # 512 → 1024

        # Decoder
        self.u1  = Up(base_ch * 16, base_ch * 8)           # 1024 → 512
        self.u2  = Up(base_ch * 8,  base_ch * 4)           # 512 → 256
        self.u3  = Up(base_ch * 4,  base_ch * 2)           # 256 → 128
        self.u4  = Up(base_ch * 2,  base_ch)               # 128 → 64

        # Output heads
        self.param_head   = nn.Conv2d(base_ch, 5, 1)       # 5 physical params
        self.feature_head = nn.Conv2d(base_ch, base_ch, 1) # features for FNO

    def forward(self, x):
        # Encoder path (save skip connections)
        s0 = self.inc(x)    # (B, 64,   128, 128)
        s1 = self.d1(s0)    # (B, 128,   64,  64)
        s2 = self.d2(s1)    # (B, 256,   32,  32)
        s3 = self.d3(s2)    # (B, 512,   16,  16)
        b  = self.bot(s3)   # (B, 1024,   8,   8)

        # Decoder path (use skip connections)
        d  = self.u1(b,  s3)  # (B, 512,  16,  16)
        d  = self.u2(d,  s2)  # (B, 256,  32,  32)
        d  = self.u3(d,  s1)  # (B, 128,  64,  64)
        d  = self.u4(d,  s0)  # (B, 64,  128, 128)

        # Physical parameter maps
        raw   = self.param_head(d)              # (B, 5, 128, 128)
        K     = F.softplus(raw[:, 0:1])         # hydraulic conductivity > 0
        c     = F.softplus(raw[:, 1:2])         # cohesion > 0
        phi   = torch.sigmoid(raw[:, 2:3]) * 45.0  # friction angle 0-45
        n     = torch.sigmoid(raw[:, 3:4])      # porosity 0-1
        alpha = F.softplus(raw[:, 4:5])         # van Genuchten > 0

        params   = torch.cat([K, c, phi, n, alpha], dim=1)  # (B, 5, 128, 128)
        features = self.feature_head(d)                      # (B, 64, 128, 128)

        return params, features


class SpectralConv2d(nn.Module):
    def __init__(self, in_channels, out_channels, modes):
        super().__init__()
        self.in_channels  = in_channels
        self.out_channels = out_channels
        self.modes        = modes

        scale = 1.0 / (in_channels * out_channels)

        # Learnable complex weights in frequency domain
        self.weights1 = nn.Parameter(
            scale * torch.rand(in_channels, out_channels, modes, modes,
                               dtype=torch.cfloat))
        self.weights2 = nn.Parameter(
            scale * torch.rand(in_channels, out_channels, modes, modes,
                               dtype=torch.cfloat))

    def compl_mul2d(self, inp, weights):
        return torch.einsum("bixy,ioxy->boxy", inp, weights)

    def forward(self, x):
        B, C, H, W = x.shape

        # Forward FFT+
        x_ft = torch.fft.rfft2(x)

        # Multiply Fourier modes with learned weights
        out_ft = torch.zeros(B, self.out_channels, H, W // 2 + 1,
                             dtype=torch.cfloat, device=x.device)

        m = self.modes
        out_ft[:, :, :m, :m]  = self.compl_mul2d(x_ft[:, :, :m, :m],  self.weights1)
        out_ft[:, :, -m:, :m] = self.compl_mul2d(x_ft[:, :, -m:, :m], self.weights2)

        # Inverse FFT
        x_out = torch.fft.irfft2(out_ft, s=(H, W))
        return x_out


class FNOBlock(nn.Module):
    def __init__(self, width, modes):
        super().__init__()
        self.spectral = SpectralConv2d(width, width, modes)
        self.W        = nn.Conv2d(width, width, 1)

    def forward(self, x):
        return F.gelu(self.spectral(x) + self.W(x))


class FourierNeuralOperator(nn.Module):
    def __init__(self, in_channels, width=32, modes=16, n_layers=4):
        super().__init__()
        self.lift   = nn.Conv2d(in_channels, width, 1)
        self.blocks = nn.ModuleList(
            [FNOBlock(width, modes) for _ in range(n_layers)]
        )
        self.proj   = nn.Sequential(
            nn.Conv2d(width, width * 2, 1),
            nn.GELU(),
            nn.Conv2d(width * 2, 2, 1),  # output: pore pressure + FoS
        )

    def forward(self, x):
        x = self.lift(x)
        for block in self.blocks:
            x = block(x)
        return self.proj(x)


class LSTF_PDE(nn.Module):
    def __init__(self, in_channels=14, base_ch=64,
                 fno_modes=16, fno_width=32, fno_layers=4):
        super().__init__()

        # Stage 1: Parameter Field Generator (U-Net)
        self.pfg = ParameterFieldGenerator(
            in_channels=in_channels,
            base_ch=base_ch
        )

        # Stage 2: Fourier Neural Operator (PDE Solver)
        fno_in   = base_ch + 5        # 64 features + 5 param maps = 69
        self.fno = FourierNeuralOperator(
            in_channels=fno_in,
            width=fno_width,
            modes=fno_modes,
            n_layers=fno_layers
        )

        # Stage 3: Failure Detection Head
        self.failure_head = nn.Sequential(
            nn.Conv2d(2 + 5 + base_ch, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.GELU(),
            nn.Conv2d(64, 32, 3, padding=1),
            nn.GELU(),
            nn.Conv2d(32, 1, 1),      # final output: landslide probability
        )

    def forward(self, x):
        # Stage 1: U-Net generates physical parameter maps
        param_maps, features = self.pfg(x)
        # param_maps : (B, 5,  128, 128)
        # features   : (B, 64, 128, 128)

        # Stage 2: FNO solves coupled PDEs
        fno_input = torch.cat([features, param_maps], dim=1)  # (B, 69, 128, 128)
        pde_out   = self.fno(fno_input)                        # (B, 2,  128, 128)

        pore_map = pde_out[:, 0:1]                             # pore water pressure
        fos_map  = F.softplus(pde_out[:, 1:2]) + 0.01         # Factor of Safety

        # Stage 3: Failure detection
        head_input = torch.cat([pore_map, fos_map,
                                param_maps, features], dim=1)  # (B, 71, 128, 128)
        logits = self.failure_head(head_input)                  # (B, 1,  128, 128)

        return logits, fos_map, pore_map, param_maps, features



class DiceLoss(nn.Module):
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth

    def forward(self, logits, targets):
        probs   = torch.sigmoid(logits).squeeze(1)  # (B, H, W)
        targets = targets.float()

        inter = (probs * targets).sum(dim=(1, 2))
        union = probs.sum(dim=(1, 2)) + targets.sum(dim=(1, 2))
        dice  = (2.0 * inter + self.smooth) / (union + self.smooth)

        return 1.0 - dice.mean()


class PDEResidualLoss(nn.Module):
    def forward(self, pore_map, K_field):
        # Laplacian of pore pressure (finite difference)
        laplacian_kernel = torch.tensor(
            [[0,  1, 0],
             [1, -4, 1],
             [0,  1, 0]], dtype=pore_map.dtype, device=pore_map.device
        ).view(1, 1, 3, 3)

        lap_pore = F.conv2d(pore_map, laplacian_kernel, padding=1)

        # Spatial smoothness of K field
        grad_K_x = torch.diff(K_field, dim=3, prepend=K_field[:, :, :, :1])
        grad_K_y = torch.diff(K_field, dim=2, prepend=K_field[:, :, :1, :])
        grad_K_mag = (grad_K_x ** 2 + grad_K_y ** 2).mean()

        # PDE residual: K * Laplacian(pore) should be smooth
        residual = (K_field * lap_pore).pow(2).mean()

        return residual + 0.1 * grad_K_mag


    
class FoSLoss(nn.Module):
    def forward(self, fos_map, mask):
        # At landslide pixels, FoS must be <= 1.0
        # Penalize if FoS > 1.0 at known landslide locations
        ls_mask   = mask.unsqueeze(1).float()
        violation = F.relu(fos_map - 1.0)
        return (violation * ls_mask).mean()


class SmoothnessLoss(nn.Module):
    def forward(self, param_maps):
        # Parameter fields must vary smoothly across space
        grad_x = torch.diff(param_maps, dim=3, prepend=param_maps[:, :, :, :1])
        grad_y = torch.diff(param_maps, dim=2, prepend=param_maps[:, :, :1, :])
        return (grad_x ** 2 + grad_y ** 2).mean()


class LSTFPDELoss(nn.Module):
    def __init__(self, lambda_pde=0.1, lambda_fos=0.1,
                 lambda_smo=0.01, pos_weight=5.0):
        super().__init__()
        self.bce  = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight]))
        self.dice = DiceLoss()
        self.pde  = PDEResidualLoss()
        self.fos  = FoSLoss()
        self.smo  = SmoothnessLoss()

        self.lambda_pde = lambda_pde
        self.lambda_fos = lambda_fos
        self.lambda_smo = lambda_smo

    def forward(self, logits, fos_map, pore_map, param_maps, targets):
        targets_f = targets.float()
        self.bce.pos_weight = self.bce.pos_weight.to(logits.device)

        l_bce  = self.bce(logits.squeeze(1), targets_f)
        l_dice = self.dice(logits, targets)
        l_pde  = self.pde(pore_map, param_maps[:, 0:1])
        l_fos  = self.fos(fos_map, targets)
        l_smo  = self.smo(param_maps)

        total = (l_bce + l_dice
                 + self.lambda_pde * l_pde
                 + self.lambda_fos * l_fos
                 + self.lambda_smo * l_smo)

        return total, {
            "bce":  l_bce.item(),
            "dice": l_dice.item(),
            "pde":  l_pde.item(),
            "fos":  l_fos.item(),
            "smo":  l_smo.item()
        }



# training


device = 'cuda' if torch.cuda.is_available() else 'cpu'
m11 = LSTF_PDE(
    in_channels=14,
    base_ch=64,
    fno_modes=16,
    fno_width=32,
    fno_layers=4
).to(device)

criterion = LSTFPDELoss(
    lambda_pde=0.1, lambda_fos=0.1, lambda_smo=0.01, pos_weight=5.0
).to(device)



m11, r11, h11 = run_training(m11, 'PILL', epochs=EP, lr=LR, criterion=criterion)
ALL_RESULTS.append(r11)
ALL_HISTORIES['PILL'] = h11
torch.save(m11.state_dict(), CHECKPOINTS_DIR / "Method11_PILL.pth")
print(f"\n{'='*50}\nMethod 11 (PILL) training complete. Model saved.\n F1: {r11['F1']} IoU: {r11['IoU']} AUC: {r11['AUC-ROC']} \n{'='*50}")

```
